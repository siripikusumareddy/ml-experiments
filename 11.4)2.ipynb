```python
import pandas as pd
import numpy as np
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler
```

```python
data = {
    'Climate': [0.190, 0.017, 0.207],
    'Housing': [0.544, 0.020, 0.204],
    'Health': [0.782, -0.605, 0.144],
    'Crime': [0.365, 0.294, 0.585],
    'Transportation': [0.585, 0.085, 0.234],
    'Education': [0.394, -0.273, 0.027],
    'Arts': [0.985, 0.126, -0.111],
    'Recreation': [0.520, 0.402, 0.519],
    'Economy': [0.142, 0.150, 0.239]
}

df = pd.DataFrame(data, index=['PC1', 'PC2', 'PC3'])
df
```

```python
print("PCA Loading Matrix:")
print(df)
```

```python
for pc in df.index:
    variable = df.loc[pc].abs().idxmax()
    loading = df.loc[pc, variable]
    print(f"{pc}: {variable}, Loading = {loading}")
```

```python
X = df.T

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

print("Scaled data shape:", X_scaled.shape)
```

```python
pca = PCA(n_components=3)
pca.fit(X_scaled)

print("PCA model trained successfully")
```

```python
print("Explained Variance Ratio:")

for i, variance in enumerate(pca.explained_variance_ratio_, 1):
    print(f"PC{i}: {variance:.4f}")
```

```python
print("Conclusion:")
print("PC1: Arts has the highest loading (0.985)")
print("PC2: Health has the highest absolute loading (-0.605)")
print("PC3: Crime has the highest loading (0.585)")
```
