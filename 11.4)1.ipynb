```python

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.DataFrame({
    'Price_Lakhs': [50, 80, 120, 150, 60, 100, 200, 90, 180, 70],
    'Area_sqft': [1000, 1200, 1500, 1800, 900, 1300, 2200, 1100, 2000, 950],
    'Age_Years': [2, 5, 10, 15, 3, 8, 20, 6, 18, 4],
    'Location_Score': [8, 7, 5, 4, 9, 6, 3, 7, 4, 8],
    'Days_On_Market': [30, 100, 220, 250, 45, 190, 300, 80, 280, 50]
})
```

```python
df['Unsold_6_Months'] = df['Days_On_Market'] > 180

print("Properties unsold for more than 6 months:")
print(df[df['Unsold_6_Months']])
```

```python
print("\nAverage characteristics:")
print(df.groupby('Unsold_6_Months')[
    ['Price_Lakhs', 'Area_sqft', 'Age_Years',
     'Location_Score', 'Days_On_Market']
].mean().round(2))
```

```python
sns.scatterplot(data=df, x='Price_Lakhs',
                y='Days_On_Market', hue='Unsold_6_Months')
plt.axhline(180, linestyle='--')
plt.title('Property Price vs Days on Market')
plt.show()
```

```python
sns.heatmap(df.corr(numeric_only=True),
            annot=True, cmap='coolwarm', fmt='.2f')
plt.title('Real-Estate Feature Correlation')
plt.show()
```

```python
unsold = df[df['Days_On_Market'] > 180]

print("Properties unsold for more than 6 months:")
print(unsold)
```

```python
print(df.groupby('Unsold_6_Months')[
    ['Price_Lakhs', 'Area_sqft', 'Age_Years',
     'Location_Score', 'Days_On_Market']
].mean().round(2))
```
