```python
import pandas as pd
from sklearn.tree import DecisionTreeClassifier, plot_tree
import matplotlib.pyplot as plt

data = {
    'Age': [25, 35, 45, 32, 28, 50, 40, 22, 38, 31],
    'Income': [30, 60, 80, 40, 35, 90, 70, 25, 65, 55],
    'Buys_House': ['No', 'Yes', 'Yes', 'Yes', 'No',
                   'Yes', 'Yes', 'No', 'Yes', 'Yes']
}

df = pd.DataFrame(data)
print(df)
```

```python
X = df[['Age', 'Income']]
y = df['Buys_House']
```

```python
model = DecisionTreeClassifier(
    criterion='gini',
    max_depth=2,
    random_state=42
)

model.fit(X, y)
print("Model trained successfully")
```

```python
plt.figure(figsize=(12, 6))

plot_tree(
    model,
    feature_names=['Age', 'Income'],
    class_names=model.classes_,
    filled=True,
    rounded=True
)

plt.show()
```

```python
person = pd.DataFrame({
    'Age': [35],
    'Income': [60]
})

prediction = model.predict(person)[0]

print("Prediction:", prediction)
```

```python
probability = model.predict_proba(person)[0]

for label, prob in zip(model.classes_, probability):
    print(f"{label}: {prob * 100:.2f}%")
```
